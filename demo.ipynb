# simple_agents_framework demo

Markdown file in, agent out. `ask()` returns a string; `ask_html()` streams the
run into the cell — the agent's answer types itself out, tool calls and results
appear as they happen, color-coded per event kind (legend in the README).


```python
import re, pathlib
import simple_agents_framework as saf

ENV = pathlib.Path.home() / "Documents/Code/Shared.env"
API_KEY = re.search(r'^\s*(?:export\s+)?ANTHROPIC_API_KEY\s*=\s*["\']?([^"\'\s]+)',
                    ENV.read_text(), re.M).group(1)
print("key loaded:", API_KEY[:12] + "...")
```

## The agent definition

Frontmatter sets name/model/tools; the body is the system prompt.

```python
print(pathlib.Path("doc_reader.md").read_text())
```

```python
# ponytail: a wrapper `claude` may shadow the real CLI on PATH (see README).
CLI = pathlib.Path.home() / ".local/bin/claude"
agent = saf.create_agent_from_markdown("doc_reader.md", API_KEY,
                                       **({"cli_path": CLI} if CLI.exists() else {}))
agent, agent.options.model, agent.options.allowed_tools
```

## Streamed run

Tool calls and results appear as they happen.

```python
agent.ask_html("Which markdown files are in this folder, and what does each one specify?")
```

## Follow-up

The session resumes, so the agent still has the earlier context.

```python
agent.ask_html("What tools is that agent allowed to use, and why those?")
```

## Plain string, no HTML

```python
print(agent.ask("In one line: how many markdown files did you read?"))
```
